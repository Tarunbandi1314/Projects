### Standardize text value 

```python
# Standardize
gender_map = {
    "male": "Male",
    "m": "Male",
    "female": "Female",
    "f": "Female"
}
df["Gender"] = df["Gender"].str.lower().map(gender_map)

print(df)
```

### Clean column names used lowercase and no space

```python
df.columns = df.columns.str.strip().str.lower().str.replace(" ", "_")

print(df)
```

### Clean column names used Uppercase and no space

```python
df.columns = df.columns.str.strip().str.upper().str.replace(" ", "_")

print(df)
```
