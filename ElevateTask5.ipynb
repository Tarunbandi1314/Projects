#### This chart shows the distribution of Titanic passengers by PassengerId, separated by survival status.

```python
fig, axs = plt.subplots(1, 3, figsize=(10, 3), constrained_layout=True)
fig.suptitle("Bivariate Visualization with Seaborn")

axs[0].set_title("Scatter Plot")
sns.scatterplot(data=df, x='Survived', y='PassengerId', ax=axs[0])

axs[1].set_title("Line Plot")
sns.lineplot(data=df, x='Survived', y='PassengerId', ax=axs[1])
axs[1].annotate("What ?", xytext=(10, 6), color="red", 
                xy=(7, 3.2), arrowprops={"arrowstyle":"->", "connectionstyle":"arc3, rad=0.3"})

axs[2].set_title("Heat Map")
sns.heatmap(df.select_dtypes(include=['int64', 'float64']).corr(), annot=True, linewidths=0.8, ax=axs[2])

plt.show()
```

#### 
The scatter plot shows that PassengerId values are evenly distributed across both survived (1) and not survived (0) categories, without any visible pattern.

The line plot suggests a very slight negative trend between PassengerId and survival, but it’s almost flat.

The heat map confirms this, showing a near-zero correlation (−0.023) between PassengerId and survival

```python
def discrete_univariate_analysis(discrete_data):
    for col_name in discrete_data:
        print("*"*10, col_name, "*"*10)
        print(discrete_data[col_name].agg(['count', 'nunique', 'unique']))
        print('Value Counts: \n', discrete_data[col_name].value_counts())
        print()
```

```python
def numerical_univariate_analysis(numerical_data):
    for col_name in numerical_data:
        print("*"*10, col_name, "*"*10)
        print(numerical_data[col_name].agg(['count', 'min', 'max', 'mean', 'median', 'std', 'skew']))
        print()
```

```python
discrete_df = df.select_dtypes(include=['int64'])

numerical_df = df.select_dtypes(include=['float64', 'int64'])
```

```python
sns.histplot(data=df, x="Age")
```

#### Observations:

Age concentration:

The majority of individuals fall between the ages of 20 to 30, with a clear peak around the 20–25 range.

This suggests a younger population dominates the dataset.

Skewness:

The distribution is right-skewed (positively skewed), as the frequency gradually decreases with higher ages.

Less representation in older ages:

After age 40, the counts drop steadily, with very few individuals above 60 years.

Ages above 70 have minimal presence.

Key insight:

The dataset likely represents a population group such as students, early professionals, or a community with more young adults than older individuals.

```python
sns.boxplot(data=df, x="Age")
```

#### Observations:

Median (central tendency):

The median age lies around 28–30 years, indicating that half of the population is younger than this age and half is older.

Interquartile range (IQR):

The middle 50% of the data (Q1 to Q3) lies between approximately 22 and 38 years, showing that most individuals are within this range.

Spread of data:

The overall data ranges roughly from 0 to 65 years, suggesting a fairly wide age distribution.

Outliers:

Ages above 65–70 are considered outliers (dots on the right side). These represent a few older individuals who are not part of the main age group.

Symmetry/Skewness:

The box is slightly longer on the right, with outliers on the higher side → indicating a right-skewed distribution (consistent with the histogram you shared earlier).
